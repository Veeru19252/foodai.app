## 1. Import pandas

`pd` is the standard short name everyone uses.

```python
import pandas as pd
print("pandas version:", pd.__version__)
```

## 2. Create a DataFrame (like a table)

A DataFrame is a table with rows and columns. You can make one from a Python dictionary.

```python
data = {
    "order_id": [1, 2, 3, 4, 5],
    "zone": ["A", "B", "A", "C", "B"],
    "distance_km": [2.1, 5.4, 1.8, 7.2, 3.9],
    "delivery_min": [18, 32, 15, 41, 26],
}
df = pd.DataFrame(data)
df
```

## 3. Look at your data

- `.head()` — first rows
- `.info()` — columns + types
- `.describe()` — statistics (count, mean, min, max)

```python
print(df.head())
print()
print(df.info())
print()
print(df.describe())
```

## 4. Filter rows (this is WHERE in SQL)

Keep only rows where a condition is True.

```python
long_orders = df[df["distance_km"] > 3.0]
print("Orders longer than 3 km:")
print(long_orders)
```

## 5. Group by (this is GROUP BY in SQL)

This is THE skill for your demand forecasting. Group orders by zone and get statistics per zone.

```python
zone_stats = df.groupby("zone").agg(
    order_count=("order_id", "count"),
    avg_distance=("distance_km", "mean"),
    avg_delivery=("delivery_min", "mean"),
)
print(zone_stats)
```

## 6. Missing values

Real data has gaps. Check and handle them.

```python
import numpy as np
df_missing = pd.DataFrame({"a": [1, np.nan, 3], "b": [4, 5, 6]})
print("Is NaN?")
print(df_missing.isna())
print("\nFilled with 0:")
print(df_missing.fillna(0))
```
